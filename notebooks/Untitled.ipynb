```python
import pandas as pd
import glob as glob
```

```python
df = pd.read_csv("../data/raw/2025-01-metropolitan-street.csv")
```

```python
files = glob.glob("../data/raw/*.csv")
for f in files:
    cols = pd.read_csv(f, nrows=0).columns.tolist()
    print(f, cols == df.columns.tolist())
```

```python
list_df = [pd.read_csv(f) for f in files]
df_all = pd.concat(list_df, ignore_index=True)
```

```python
df_all.shape
```

```python
df_all["Month"] = pd.to_datetime(df_all["Month"])
print("Success")
```

```python
df_all = df_all.drop(columns=["Context"])
```

```python
df_all.duplicated().sum()
```

```python
df_all[df_all.duplicated(keep=False)]['Crime type'].value_counts()
```

```python
has_id = df_all['Crime ID'].notnull()
df_with_id = df_all[has_id].drop_duplicates()
df_without_id = df_all[~has_id]

df_all_clean = pd.concat([df_with_id, df_without_id], ignore_index=True)
print(df_all.shape, "->", df_all_clean.shape)
```

```python
df_all_clean.to_csv("../data/processed/london_crime_2025_clean.csv", index=False)
```
